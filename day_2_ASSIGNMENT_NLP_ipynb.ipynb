```python
import pandas as pd
import re
import nltk
```

```python
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer

nltk.download('stopwords')
```

```python
path = "/root/.cache/kagglehub/datasets/ibrahimqasimi/imdb-50k-cleaned-movie-reviews/versions/1/IMDB_cleaned.csv"
df = pd.read_csv(path)
```

```python
texts = df['review'].head(1000)
print(df.head(1000))
```

```python
texts = df['review'].head(10)
```

```python
stop_words=set(stopwords.words('english'))
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    words = text.split()
    words = [w for w in words if w not in stop_words]
    return " ".join(words)
    print(words)
```

```python
cleaned_texts = texts.apply(clean_text)
vectorizer = CountVectorizer()
X_bow = vectorizer.fit_transform(cleaned_texts)
```

```python
vocab = vectorizer.get_feature_names_out()

print("Vocabulary size:", len(vocab))
print("Sample words:", vocab[:20])
```

TF-IDF + Comparison



```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(cleaned_texts)

print("BoW shape:", X_bow.shape)
print("TF-IDF shape:", X_tfidf.shape)

# Compare first document
print("\nBoW vector (first doc):")
print(X_bow[0].toarray()[0][:20])

print("\nTF-IDF vector (first doc):")
print(X_tfidf[0].toarray()[0][:20])
```

```python
ngram_vectorizer = CountVectorizer(ngram_range=(1,2))
X_ngram = ngram_vectorizer.fit_transform(cleaned_texts)

ngram_features = ngram_vectorizer.get_feature_names_out()

print("N-gram feature count:", len(ngram_features))
print("Sample features:", ngram_features[:20])
```
