```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
text = """Narendra Modi is the Prime Minister of India.
He is leading the country and working on development projects."""
```

```python
import re
clean_text=re.sub('[^a-zA-Z]', ' ', text)
clean_text = clean_text.lower()

print(clean_text)
```

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt_tab')
tokens = word_tokenize(clean_text)
print(tokens)
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

filtered_words = [w for w in tokens if w not in stop_words]
print(filtered_words)
```

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

stemmed_words = [ps.stem(w) for w in filtered_words]
print(stemmed_words)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# list of documents
corpus = [clean_text]

cv = CountVectorizer()

X = cv.fit_transform(corpus)

print("Vocabulary:", cv.get_feature_names_out())
print("BoW Matrix:\n", X.toarray())
```

```python
cv_binary = CountVectorizer(binary=True)

X_bin = cv_binary.fit_transform(corpus)

print("Binary BoW:\n", X_bin.toarray())
```

```python
cv_stop = CountVectorizer(stop_words='english')

X_stop = cv_stop.fit_transform(corpus)

print("Without Stop Words:\n", X_stop.toarray())
```
