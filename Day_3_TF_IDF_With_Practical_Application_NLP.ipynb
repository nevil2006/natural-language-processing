```python
import re
```

```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
```

```python
corpus=[
    "He is a good boy",
    "She is a good girl",
    "Boy and girl are good"
]
```

```python
cleaned=[]
for text in corpus:
  text=re.sub('[^a-zA-z]',' ',text)
  text=text.lower()
  cleaned.append(text)
  print(cleaned)
```

```python
Stop_words=set(stopwords.words('english'))
```

```python
print(Stop_words)
```

```python
processed = []
for text in cleaned:
    words = text.split()
    words = [w for w in words if w not in Stop_words]
    processed.append(" ".join(words))
    print(processed)
```

```python
stop_words=Stop_words
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
X = cv.fit_transform(processed)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
print("\nBoW Vocabulary:", cv.get_feature_names_out())
print("BoW Matrix:\n", X.toarray())

# TF-IDF
tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(processed)

print("\nTF-IDF Vocabulary:", tfidf.get_feature_names_out())
print("TF-IDF Matrix:\n", X_tfidf.toarray())
```

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

stemmed = []
for text in processed:
    words = text.split()
    words = [ps.stem(w) for w in words]
    stemmed.append(" ".join(words))

print("Stemmed:", stemmed)
```

```python
cv_ngram = CountVectorizer(ngram_range=(2,2))
X_ng = cv_ngram.fit_transform(processed)

print("Bigrams:", cv_ngram.get_feature_names_out())
```
