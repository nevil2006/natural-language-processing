```python
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
import nltk

# Download tokenizer
nltk.download('punkt')

# Sample sentences
data = [
    "cat eats food",
    "dog eats food",
    "cat likes milk",
    "dog likes bone",
    "bird eats seeds",
    "cow drinks water",
    "tiger eats meat",
    "rabbit likes carrot",
    "monkey eats banana",
    "goat eats grass"
]

# Tokenize
tokenized = [word_tokenize(sentence.lower()) for sentence in data]

# Train Word2Vec model
model = Word2Vec(
    sentences=tokenized,
    vector_size=50,
    window=2,
    min_count=1,
    sg=1,        # skip-gram
    epochs=200,
    seed=42
)

print("Similar words to 'cat':")
print(model.wv.most_similar("cat"))

print("\nSimilar words to 'dog':")
print(model.wv.most_similar("dog"))

print("\nSimilarity scores:")
print("cat vs dog:", model.wv.similarity("cat", "dog"))
print("cat vs milk:", model.wv.similarity("cat", "milk"))

print("\nVector for 'cat':")
print(model.wv["cat"])

print("\nVocabulary:")
print(model.wv.index_to_key)
```

```python
data.extend([
    "this is good",
    "good food is tasty",
    "dog is good",
    "cat is good"
])
```

```python
from nltk.tokenize import word_tokenize

tokenized = [word_tokenize(sentence.lower()) for sentence in data]

model = Word2Vec(
    sentences=tokenized,
    vector_size=50,
    window=2,
    min_count=1,
    sg=1,
    epochs=200,
    seed=42
)

print(model.wv.most_similar("good"))
```
