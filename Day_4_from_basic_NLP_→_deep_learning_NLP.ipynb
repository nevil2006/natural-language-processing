```python
from gensim.models import Word2Vec
from nltk.tokenize import word_tokenize
```

```python
data = [
    "cat eats food",
    "dog eats food",
    "cat likes milk",
    "dog likes bone"
]
```

```python
import nltk
nltk.download('punkt_tab')
tokenized = [word_tokenize(sentence.lower()) for sentence in data]
```

```python
model = Word2Vec(sentences=tokenized, vector_size=50, window=2, min_count=1)

# Check similarity
print(model.wv.most_similar("cat"))

# Vector
```

```python
print(model.wv["cat"])
```

```python
from sklearn.metrics.pairwise import cosine_similarity

vec1 = model.wv["cat"].reshape(1, -1)
vec2 = model.wv["dog"].reshape(1, -1)

print(cosine_similarity(vec1, vec2))
```
